```python
from NeuroSketch.engine.nn import Linear, Sequential
from NeuroSketch.engine.act import Sigmoid, Softmax, ReLU
from NeuroSketch.utils import DataLoader
from NeuroSketch.losses import BinaryCrossentropyLoss, SparseCategoricalCrossentropyLoss
from NeuroSketch.optims import ADAM, SGD
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('fashion-mnist_test.csv')
df = df.dropna()
```

```python
df.head()
```

```python
X = df.drop(columns=["label"]).to_numpy()
y = df['label'].to_numpy()
```

```python
y.shape
```

```python
n = len(y)
x_train, x_test = X[:n // 2], X[n // 2:]
y_train, y_test = y[:n // 2], y[n // 2:]
```

```python
x_test.shape
```

```python
train = DataLoader(x_train, y_train, shuffle=False)
```

```python
np.unique(y)
```

```python
test = DataLoader(x_test, y_test)
```

```python
model = Sequential(
    Linear(784,10),
    Softmax()
)
```

```python
criterion = SparseCategoricalCrossentropyLoss(model.layers[-1])
```

```python
optim = ADAM(model, lr = 0.01)
```

```python
for epoch in range(100):
    for x_b, y_b in train():
        pred = model(x_b)
        loss = criterion(pred, y_b)
        # print(f"epoch: {epoch} loss: {loss}")
        criterion.backward()
        optim.step()
```

```python
x_t = test()[0][0]
y_t = test()[0][1]
```

```python
x_t.shape
```

```python
loss
```

```python
y_pred = model(x_t)
labeled = np.argmax(y_pred, axis=1).reshape(-1, 1)
accuracy = np.mean(labeled == y_t)
print(f"Test accuracy: {accuracy:.4f}")
```
